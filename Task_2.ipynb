```python
# Install libraries silently
!pip install -q SpeechRecognition pydub

# Install ffmpeg silently
!apt-get -qq install ffmpeg

import speech_recognition as sr
from pydub import AudioSegment
import os

def convert_audio():

    print("\n====================================")
    print("      AUDIO CONVERSION STARTED")
    print("====================================")

    # Load audio file
    audio = AudioSegment.from_file("Record.m4a")

    # Export as WAV
    audio.export("converted.wav", format="wav")

    # Calculate duration
    duration = len(audio) / 1000

    print("\nAudio converted successfully!")
    print(f"Audio Duration: {duration:.2f} seconds")

    return "converted.wav"

def speech_to_text(audio_file):

    recognizer = sr.Recognizer()

    try:

        # Open audio file
        with sr.AudioFile(audio_file) as source:

            print("\n====================================")
            print("        PROCESSING AUDIO")
            print("====================================")

            # Read audio data
            audio_data = recognizer.record(source)

            print("\nRecognizing speech...\n")

            # Convert speech into text
            text = recognizer.recognize_google(audio_data)

            print("====================================")
            print("         TRANSCRIBED TEXT")
            print("====================================")

            print(text)

            print("====================================")

            # Save output to text file
            with open("Speech_Output.txt", "w") as file:

                file.write("===== SPEECH RECOGNITION OUTPUT =====\n\n")
                file.write(text)

            print("\nText saved successfully!")
            print("File Name: Speech_Output.txt")

    except sr.UnknownValueError:

        print("Could not understand the audio.")

    except sr.RequestError as e:

        print(f"API Error: {e}")

    except FileNotFoundError:

        print("Audio file not found.")

    except Exception as e:

        print(f"Error: {e}")

if __name__ == "__main__":

    print("====================================")
    print("     SPEECH RECOGNITION SYSTEM")
    print("====================================")

    # Check if file exists
    if os.path.exists("Record.m4a"):

        # Convert audio
        wav_file = convert_audio()

        # Speech recognition
        speech_to_text(wav_file)

    else:

        print("\nERROR: 'Record.m4a' file not found.")
        print("Please upload the audio file first.")
```
